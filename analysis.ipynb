```python
crop_season_performance = df.groupby(
    ["Season", "Crop"]
)["Yield_Tonnes_Ha"].mean()


# Find the crop with the highest yield in each season

best_crop_yield = crop_season_performance.groupby(
    level=0
).idxmax()


print("Best Crop by Yield in Each Season:")

for season, crop in best_crop_yield.items():
    print(season, ":", crop[1])
```

```python
crop_season_profit_group = df.groupby(
    ["Season", "Crop"]
)["Profit_INR"].mean()


# Find the most profitable crop in each season

best_crop_profit = crop_season_profit_group.groupby(
    level=0
).idxmax()


print("Most Profitable Crop in Each Season:")

for season, crop in best_crop_profit.items():
    print(season, ":", crop[1])
```
